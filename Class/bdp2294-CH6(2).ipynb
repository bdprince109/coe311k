Jacobi Algorithm

```python
import numpy as np
def jacobi(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]
        # Check for convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k
        x = x_new
    return x, max_iterations
```

Gauss-Seidel Algorithm

```python
import numpy as np
def gauss_seidel(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = x.copy()
        for i in range(n):
            s1 = sum(A[i][j] * x_new[j] for j in range(i)) # Using already updated values
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n)) # Using old values
            x_new[i] = (b[i] - s1 - s2) / A[i][i]
        # Check for convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k
        x = x_new
    return x, max_iterations
```

Challenge 6a.

```python
import time
A=np.array([[10, -1, 2, 0, 0],[-1, 11, -1, 3, 0],[2, -1, 10, -1, 0],[0, 3, -1, 8, -2],[0, 0, 0, -2, 9]])
b=np.array([14.0, 30.0, 26.0, 25.0, 37.0])
x0 = np.zeros_like(b)
tol = 1e-6
max_iterations = 100
t1=time.time()
solution, iterations = jacobi(A, b, x0, tol, max_iterations)
print(f"Solution: {solution}")
print(f"Iterations: {iterations}")
t2=time.time()
print(t2-t1)

#gauss
t3=time.time()
solution, iterations = gauss_seidel(A, b, x0, tol, max_iterations)
print(f"Solution: {solution}")
print(f"Iterations: {iterations}")
t4=time.time()
print(t4-t3)
```

In this case both the Jacobi and Gauss-Seidel methods converged, however, the Jacobi method converged in 18 iterations while the Gauss-Seidel method converged in 10 iterations. The Jacobi method took 0.001374959945678711s, while the Gauss-Seidel method took  0.0015935897827148438s.
