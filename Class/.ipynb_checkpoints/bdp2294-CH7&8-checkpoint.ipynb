## Jacobi

```python
import numpy as np
def jacobi(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]
        # Check for convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k
        x = x_new
    return x, max_iterations
```

## Gauss-Seidel

```python
import numpy as np
def gauss_seidel(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = x.copy()
        for i in range(n):
            s1 = sum(A[i][j] * x_new[j] for j in range(i)) # Using already updated values
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n)) # Using old values
            x_new[i] = (b[i] - s1 - s2) / A[i][i]
        # Check for convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k
        x = x_new
    return x, max_iterations
```

## Diagonal Dominance Function

```python
def diagonal(A):
    rows=A.shape[0]
    cols=A.shape[1]
    for i in range(0,rows):
        sum=0
        for j in range(0,cols):
            if (i==j):
                continue
            sum=sum+A[i,j]
            if (A[i,i]<sum):
                print("Matrix is not diagonally dominant")
                return "False"
    return "True"
                
```

## A

```python
import time
A = np.array([[0.582745, 0.48, 0.10, 0., 0.],[0.48, 1.044129, 0.46, 0.10, 0.],[0.10, 0.46, 1.10431 , 0.44, 0.10],[0., 0.10, 0.44, 0.963889, 0.42],[0., 0., 0.10, 0.42, 0.522565]], dtype=float)
b = np.array([1.162745, 2.084129, 2.20431 , 1.923889, 1.042565], dtype=float)
x0 = np.zeros_like(b)
tol = 1e-6
max_iterations = 100
t1=time.time()
solution, iterations = jacobi(A, b, x0, tol, max_iterations)
print(f"Solution: {solution}")
print(f"Iterations: {iterations}")
t2=time.time()
print(t2-t1)

#gauss
t3=time.time()
solution, iterations = gauss_seidel(A, b, x0, tol, max_iterations)
print(f"Solution: {solution}")
print(f"Iterations: {iterations}")
t4=time.time()
print(t4-t3)
```
