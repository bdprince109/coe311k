```python
import numpy as np
import matplotlib.pyplot as plt
# Parameters
h_val = 1000.0
T0 = 500.0
t0 = 0.0
tf = 20.0
def T_ambient(t):
    return 300 + 50 * np.sin(0.01 * t)
# functions
def f_nonlinear(t, T):
    return-h_val * np.exp(T- T_ambient(t))
def df_nonlinear(t, T):
    return-h_val 
# Basic Newton-Raphson
def basic_newton_raphson(g, dg, x0, tol=1e-6, max_iter=50):
    x = x0
    for i in range(max_iter):
        try:
            x_new = x- g(x) / dg(x)
        except:
            return None, i+1, False
        if np.isnan(x_new) or abs(x_new) > 1e6:
            return None, i+1, False
        if abs(x_new- x) < tol:
            return x_new, i + 1, True
        x = x_new
    return None, max_iter, False
# Damped Newton-Raphson
def damped_newton_raphson(g, dg, x0, tol=1e-6, max_iter=50, alpha_min=1e-4):
    x = x0
    total_backtracks = 0
    final_alpha = 1.0
    damping_needed = False
    nr_count=0
    for i in range(max_iter):
        nr_count=nr_count+1
        g_val = g(x)
        if abs(g_val) < tol:
            return x, i, True, total_backtracks, final_alpha, damping_needed
        delta =-g_val / dg(x)
        alpha = 1.0
        g_current = abs(g_val)
        for _ in range(20):
            y_new = x + alpha * delta
        if abs(g(y_new)) < g_current:
            break
        alpha /= 2.0
        total_backtracks += 1
        if alpha < alpha_min:
            break
        if alpha < 1.0:
            damping_needed = True
        final_alpha = alpha
        x = x + alpha * delta
    return x, max_iter, False, total_backtracks, final_alpha, damping_needed, nr_count
# Backward Euler (Basic NR)
def backward_euler_basic(f, df, y0, t0, tf, h):
    t_values = np.arange(t0, tf + h, h)
    y_values = np.zeros_like(t_values)
    y_values[0] = y0
    converged = True
    for i in range(1, len(t_values)):
        t_next = t_values[i]
        T_prev = y_values[i-1]
        g = lambda T: T- T_prev- h * f(t_next, T)
        dg = lambda T: 1- h * df(t_next, T)
        result, iters, success = basic_newton_raphson(g, dg, x0=T_prev)
        if not success or result is None:
            converged = False
            y_values[i] = np.nan
        else:
            y_values[i] = result
    return t_values, y_values, converged
# Backward Euler (Damped NR)
def backward_euler_damped(f, df, y0, t0, tf, h):
    t_values = np.arange(t0, tf + h, h)
    y_values = np.zeros_like(t_values)
    y_values[0] = y0
    iterations_log = []
    backtrack_log = []
    alpha_log = []
    damping_used_log = []
    for i in range(1, len(t_values)):
        t_next = t_values[i]
        T_prev = y_values[i-1]
        g = lambda T: T- T_prev- h * f(t_next, T)
        dg = lambda T: 1- h * df(t_next, T)
        result, max_iters, success, backtracks, final_alpha, damping_needed,nr_count = \
            damped_newton_raphson(g, dg, x0=T_prev)
        y_values[i] = result
        iterations_log.append(nr_count)
        backtrack_log.append(backtracks)
        alpha_log.append(final_alpha)
        damping_used_log.append(damping_needed)
    return t_values, y_values, iterations_log, backtrack_log, alpha_log,damping_used_log
# Run Test (Large Step Size)
h_large = 2.0
t_basic, T_basic, conv_basic = backward_euler_basic(f_nonlinear, df_nonlinear, T0, t0, tf, h_large)
t_damped, T_damped, iters, backs, alphas, damp_used = \
backward_euler_damped(f_nonlinear, df_nonlinear, T0, t0, tf, h_large)
# Plot 1: Comparison
plt.figure()
plt.plot(t_basic, T_basic, 'r--', label='Basic NR (fails)')
plt.plot(t_damped, T_damped, 'b--', label='Damped NR (succeeds)')
plt.title('Basic vs Damped Newton (Nonlinear, Large h)')
plt.xlabel('Time')
plt.ylabel('Temperature')
plt.legend()
plt.grid(True)
plt.show()
# Plot 2: Backtracking and alpha values over time
plt.figure()
3
plt.plot(t_damped[1:], backs, 'bo-', markersize=4)
plt.title('Backtracking Steps per Time Step')
plt.xlabel('Time')
plt.ylabel('Backtracking Steps')
plt.grid(True)
plt.show()
# Plot 3: Alpha values
plt.figure()
plt.plot(t_damped[1:], alphas, 'go-', markersize=4)
plt.axhline(1.0, linestyle=':')
plt.title('Final Alpha per Time Step')
plt.xlabel('Time')
plt.ylabel('Alpha')
plt.grid(True)
plt.show()
# Print Results
print("Basic NR converged:", conv_basic)
print("Damped NR used damping:", any(damp_used))
print("Total backtracking steps:", sum(backs))
print("Steps with damping:", sum(damp_used), "out of", len(damp_used))
print("Min alpha:", min(alphas))
print("Max alpha:", max(alphas))
```

```python
# --- Global Parameters ---
h_val = 1000
T0 = 500
t0 = 0
tf = 20

def T_ambient(t):
    return 300 + 50 * np.sin(0.01 * t)

def f(t, T):
    return -h_val * (T - T_ambient(t))

def f_prime(T):
    return -h_val

# --- Damped Newton-Raphson Solver ---
def damped_nr_step(f, y_prev, t_next, h, tol=1e-6, max_iter=50):
    y_curr = y_prev # Initial guess
    nr_count=0
    for i in range(max_iter):
        nr_count=nr_count+1
        # Calculate the residual g(y)
        g_val = y_curr - y_prev - h * f(t_next, y_curr)
        dg_val = 1 - h * f_prime(y_curr)
        
        # Calculate the full Newton step
        delta = -g_val / dg_val
        
        # Line search for damping factor alpha
        alpha = 1.0
        while abs(y_curr + alpha * delta - y_prev - h *
                  f(t_next, y_curr + alpha * delta)) >= abs(g_val):
            alpha = alpha / 2.0
            if alpha < 1e-4:
                break
        
        # Update with damped step
        y_new = y_curr + alpha * delta
        
        if abs(y_new - y_curr) < tol:
            return y_new, i + 1
        y_curr = y_new
        
    return y_curr, nr_count

# --- Adaptive Implicit Euler ---
def adaptive_implicit_euler(f, y0, t0, t_final, h0, tol):
    # Initialize lists to store our results
    t_list = [t0]
    y_list = [y0]
    h_list = [] # To keep track of step sizes used
    
    # Starting values
    t_current = t0
    y_current = y0
    h = h0
    
    # Counters for stats
    accepted_count = 0
    rejected_count = 0
    total_nr=0
    while t_current < t_final:
        # need the time at the end of the step
        t_next = t_current + h
        y_full, nr_count = damped_nr_step(f, y_current, t_next, h)
        total_nr=total_nr+nr_count
        h_half = h / 2.0
        # First half-step to the middle
        t_mid = t_current + h_half
        y_mid,nr_count = damped_nr_step(f, y_current, t_mid, h_half)
        total_nr=total_nr+nr_count
        # Second half-step to the end
        y_half,nr_count = damped_nr_step(f, y_mid, t_next, h_half)
        total_nr=total_nr+nr_count
        
        # absolute difference between the two methods
        error_est = abs(y_half - y_full)
        
        if error_est < tol:
            # ACCEPT
            t_current = t_current + h
            y_current = y_half
            
            t_list.append(t_current)
            y_list.append(y_current)
            h_list.append(h)
            
            accepted_count = accepted_count + 1
            
            # try to make h bigger
            h_new = h * np.sqrt(tol / (error_est + 1e-10))
            h = 0.9 * h_new
        else:
            # REJECT
            rejected_count = rejected_count + 1
            h_new = h * np.sqrt(tol / (error_est + 1e-10))
            h = 0.9 * h_new
            
        # h doesn't get too small or too big
        if h > 1.0:
            h = 1.0
        if h < 1e-6:
            h = 1e-6
            
        # don't go past the final time
        if t_current + h > t_final:
            h = t_final - t_current
    return np.array(t_list), np.array(y_list), h_list, accepted_count, rejected_count, total_nr
def backward_euler_damped1(f, y0, t0, t_final, h):
    t_values = np.arange(t0, tf + h, h)
    y_values = np.zeros_like(t_values)
    y_values[0] = y0
    iterations_log = []
    backtrack_log = []
    alpha_log = []
    damping_used_log = []
    for i in range(1, len(t_values)):
        t_next = t_values[i]
        y_prev = y_values[i-1]
        result, nr_count=damped_nr_step(f, y_prev, t_next, h, tol=1e-6, max_iter=50)
        y_values[i] = result
        iterations_log.append(nr_count)
    return t_values, y_values, iterations_log, backtrack_log, alpha_log,damping_used_log
```

## Performance Analysis

#### Fixed Solve 

```python
T0=500
t0=0
t_final=20
t_damped, T_damped, iters, backs, alphas, damp_used = \
backward_euler_damped1(f, T0, t0, t_final, 1)
print(np.sum(iters))
```

#### Adaptive Solve

```python
T0 = 500.0
t_start, t_end = 0, 20
tol_target = 1e-3
t_vals, T_vals, h_vals, accepted_count, rejected_count, total_nr \
     = adaptive_implicit_euler(f, T0, t_start, t_end, .5, tol_target)
print(total_nr)
print(accepted_count,rejected_count)
```
