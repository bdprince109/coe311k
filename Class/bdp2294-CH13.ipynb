```python
import numpy as np
import time as time
def naive_matrix_multiplication(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]
    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]
    return C
print("started")
a1=np.random.randint(0,11,size=(3,3))
a2=np.random.randint(0,11,size=(3,3))
b1=np.random.randint(0,11,size=(6,6))
b2=np.random.randint(0,11,size=(6,6))
c1=np.random.randint(0,11,size=(12,12))
c2=np.random.randint(0,11,size=(12,12))
d1=np.random.randint(0,11,size=(24,24))
d2=np.random.randint(0,11,size=(24,24))
e1=np.random.randint(0,11,size=(48,48))
e2=np.random.randint(0,11,size=(48,48))
f1=np.random.randint(0,11,size=(96,96))
f2=np.random.randint(0,11,size=(96,96))
g1=np.random.randint(0,11,size=(192,192))
g2=np.random.randint(0,11,size=(192,192))
h1=np.random.randint(0,11,size=(384,384))
h2=np.random.randint(0,11,size=(384,384))
print("done")
```

```python
t=np.zeros(9)
t[0]=time.perf_counter()
result1=np.dot(a1,a2)
t[1]=time.perf_counter()
result2=np.dot(b1,b2)
t[2]=time.perf_counter()
result3=np.dot(c1,c2)
t[3]=time.perf_counter()
result4=np.dot(d1,d2)
t[4]=time.perf_counter()
result5=np.dot(e1,e2)
t[5]=time.perf_counter()
result6=np.dot(f1,f2)
t[6]=time.perf_counter()
result7=np.dot(g1,g2)
t[7]=time.perf_counter()
result8=np.dot(h1,h2)
t[8]=time.perf_counter()
dt=np.diff(t)
print(dt)
```

```python
print("started")
tc=np.zeros(9)
tc[0]=time.perf_counter()
Cresult1=naive_matrix_multiplication(a1, a2)
tc[1]=time.perf_counter()
Cresult2=naive_matrix_multiplication(b1, b2)
tc[2]=time.perf_counter()
Cresult3=naive_matrix_multiplication(c1, c2)
tc[3]=time.perf_counter()
Cresult4=naive_matrix_multiplication(d1, d2)
tc[4]=time.perf_counter()
print("midway")
Cresult5=naive_matrix_multiplication(e1, e2)
tc[5]=time.perf_counter()
Cresult6=naive_matrix_multiplication(f1, f2)
tc[6]=time.perf_counter()
Cresult7=naive_matrix_multiplication(g1, g2)
tc[7]=time.perf_counter()
Cresult8=naive_matrix_multiplication(h1, h2)
tc[8]=time.perf_counter()
dtc=np.diff(tc)
print(dtc)
print("done")
```

```python
import matplotlib.pyplot as plt
x=np.array([3,6,12,24,48,96,192,384])
plt.figure()
plt.suptitle("Runtime vs Square Matrix Dimension")
plt.subplot(1,2,1)
plt.plot(x,dt,'r-',label="numpy")
plt.xlabel("n, nxn matrix")
plt.ylabel("seconds")
plt.legend()
plt.grid()
plt.subplot(1,2,2)
plt.plot(x,dtc,'b-',label="naive")
plt.xlabel("n, nxn matrix")
plt.ylabel("seconds")
plt.legend()
plt.grid()
plt.subplots_adjust(wspace=.25)
plt.show()
```

### Analysis
The run time for the naive function was vastly larger than the runtime for the numpy optimized version. However, both have the same shape which tells me they both grow according to O(n^3), with the time function for the optimized matrix multiplication multiplied by a smaller coefficient. This makes the time for each matrix multiplication with the optimized version smaller although it progresses with the same cubic shape.
