# Euler's Forward Solve

```python
import numpy as np
import matplotlib.pyplot as plt
t_exact=np.arange(0,5.01,.01)
analytical=3-.998*np.exp(-1000*t_exact)-2.002*np.exp(-1*t_exact) #Define analytical solution
y_0=0 #Intitial condition
string=np.array(["h=.00001","h=.0001", "h=.001", "h=.01","h=.1"])

h_array=np.array([.00001, .0001, .001,.01]) #Create array of h values
for i in range (0,h_array.size): #loop through step sizes
    h=h_array[i]
    t=np.arange(0,5+h,h)#Create array of t values
    y_values=np.zeros(t.size) #Create array of y values
    y_values[0]=y_0
    for j in range (0,t.size-1): #Implement Euler's Forward
        dy_dt=-1000*y_values[j]+3000-2000*np.exp(-1*t[j]) #Calculate derivative for the current time step
        y_values[j+1]=y_values[j]+h*dy_dt
    plt.plot(t, y_values, label=string[i])
plt.plot(t_exact, analytical, label="Exact solution")
plt.title("Euler's Forward Method for dy/dt=-1000*y+3000-2000e^-t")
plt.xlabel("T")
plt.ylabel("Y")
plt.legend()
plt.show()
        
    
```
