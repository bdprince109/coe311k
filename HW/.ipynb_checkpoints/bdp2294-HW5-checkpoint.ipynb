# Least Squares Algorthm

### Python Function

```python
import numpy as np
import matplotlib.pyplot as plt
# Data points
X = np.sort(np.random.uniform(0,10,20))
print(X)
Y = np.random.uniform(0,10,20)
# Stack the design matrix with a column of ones (for the intercept)
A = np.vstack([X, np.ones(len(X))]).T
# Solve using least squares
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
print(f"Slope: {m}, Intercept: {c}")
t=np.linspace(0,10,40)
```

```python
y=m*t+c
```

```python
plt.figure()
plt.title("Least Squares Regression")
plt.plot(X,Y, 'b.')
plt.plot(t,y,'r')
plt.show()
```

### Function Code

```python
def lstsq(X,Y):
    A=np.array([[Y.size,np.sum(X)],[np.sum(X),np.sum(X**2)]])
    rhs=np.array([np.sum(Y),np.sum(Y*X)])
    b,m=np.linalg.solve(A,rhs)
    return b,m
```

### Function Call and Plot

```python
b,m=lstsq(X,Y)
#lstsq function
t=np.linspace(0,10,40)
y=m*t+b
```

```python
plt.figure()
plt.title("Least Squares Regression")
plt.plot(X,Y, 'b.')
plt.plot(t,y,'r')
plt.show()
```

### Comparison
My least squares function gives the same result as the python least squares function.

### Linear Least Squares

```python
import numpy as np
import matplotlib.pyplot as plt
# Data points
X = np.linspace(0,1,101)
Y = 4*X+15+np.random.uniform(-1,1,101)
b,m=lstsq(X,Y)
print(b,m)
y=m*X+b
```

### Plot

```python
plt.figure()
plt.title("Least Squares Regression")
plt.plot(X,Y, 'b.')
plt.plot(X,y,'r')
plt.show()
```

### Residuals

```python
r=np.abs(Y-y)
rindex=np.arange(0,r.size,1)
print(r)
print("Minimum Error=", min(r))
print("Maximum Erorr=",max(r))
print("The minimum error occurs at X=", X[rindex[r-.0001<min(r)]])
print("The maximum error occurs at X=", X[rindex[r+.0001>max(r)]])
```

```python
plt.figure()
plt.title("Least Squares Regression")
plt.plot(X,Y, 'b.')
plt.plot(t,y,'r')
plt.show()
r=np.abs(Y-y)
rindex=np.arange(0,r.size,1)
r=np.abs(Y-y)
rindex=np.arange(0,r.size,1)
print(r)
print("Minimum Residual=", min(r))
print("Maximum Residual=",max(r))
print("The minimum residual occurs at X=", X[rindex[r-.0001<min(r)]])
print("The maximum residual occurs at X=", X[rindex[r+.0001>max(r)]])
```
