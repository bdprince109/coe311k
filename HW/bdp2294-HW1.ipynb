# Monte Carlo Function

```python
import random
import math
import numpy as np
import matplotlib.pyplot as plt
def randomShots(numSamples):
    countTotal=0
    countCircle=0
    for i in range(0,numSamples):
        x=random.uniform(-1,1)
        y=random.uniform(-1,1)
        countTotal=countTotal+1
        if (math.sqrt(x*x+y*y)<1):
            countCircle=countCircle+1
    PI=4*(countCircle/countTotal)
    return PI
```

## Error Calculation

```python
def errorCalc(PI):
    absolute=np.abs(np.pi-PI)
    relative=np.abs((np.pi-PI)/np.pi)
    return absolute, relative
```

## Plot Routine

```python
x=np.arange(10,1001,20)
absolute=np.zeros(x.size)
relative=np.zeros(x.size)
for i in range(0,x.size):
    pi=randomShots(x[i])
    absolute[i],relative[i]=errorCalc(pi)
plt.figure()
plt.plot(x,absolute,'r--',label="absolute error")
plt.plot(x,relative,'b-',label="relative error")
plt.legend()
plt.show()
    
    
```
