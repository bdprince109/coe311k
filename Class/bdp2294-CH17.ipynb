# Second Order ODE Stability Analysis

```python
import numpy as np
import matplotlib.pyplot as plt
# Define the parameters
k = 1.0 # Constant in the equation (e.g., spring constant)
h_array = np.array([0.01,.02,.03,.05,.1,.2,.3,.5]) # Step size
x_start = 0 # Start of the interval
x_end = 10 # End of the interval
y_0 = 1.0 # Initial condition for y
v_0 = 0.0 # Initial condition for dy/dx
# Create arrays to store x, y, and v values
# Set initial conditions
```

```python
plt.figure(figsize=(10, 10))
string=np.array(["Position h=.01","Velocity h=.01", "Position h=.02", "Velocity h=.02", \
                 "Position h=.03","Velocity h=.03","Position h=.05","Velocity h=.05", \
                 "Position h=.1","Velocity h=.1","Position h=.2", "Velocity h=.2", "Position h=.3", \
                 "Velocity h=.3", "Position h=.5", "Velocity h=.5"])
# Euler's Method loop
for j in range (0,h_array.size):
    h=h_array[j]
    x_values = np.arange(x_start, x_end + h, h)
    y_values = np.zeros(len(x_values))
    v_values = np.zeros(len(x_values))
    y_values[0] = y_0
    v_values[0] = v_0
    for i in range(1, len(x_values)):
        # Current values
        x_n = x_values[i - 1]
        y_n = y_values[i - 1]
        v_n = v_values[i - 1]
        # Update equations
        y_values[i] = y_n + h * v_n
        v_values[i] = v_n + h * (-k * y_n)
        # Plot the results    
    plt.plot(x_values, y_values,'-', label=string[j*2])
    plt.plot(x_values, v_values, '--', label=string[j*2+1])
plt.xlabel('x')
plt.ylabel('y and v')
plt.title('Solution of d²y/dx² = -ky using Euler\'s Method')
plt.legend()
plt.show()
```
