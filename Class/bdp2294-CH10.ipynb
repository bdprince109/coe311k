# In Class Sample Code

```python
# Importing necessary libraries
import numpy as np
import matplotlib.pyplot as plt
# Step 1: Generate some sample data
# Creating data points that roughly follow a quadratic curve
np.random.seed(0)
x = np.linspace(-5, 5, 50) # 50 points between -5 and 5
y = 2 * x**2 + 3 * x + np.random.randn(50) * 10 # y = 2x^2 + 3x + some noise
```

```python
# Step 2: Polynomial fitting using NumPy's polyfit
# Fit a 2nd degree polynomial to the data
degree = 2 # Degree of the polynomial
coefficients = np.polyfit(x, y, degree)
# Print the polynomial coefficients
print("Polynomial Coefficients: ", coefficients)
# Step 3: Use the coefficients to predict y-values (polynomial function)
# Generate polynomial function based on the coefficients
polynomial = np.poly1d(coefficients)
```

```python
# Generate predicted y-values based on the fitted polynomial
y_pred = polynomial(x)
# Step 4: Plotting the results
plt.scatter(x, y, color='blue', label='Data points') # Original data points
plt.plot(x, y_pred, color='red', label=f'Polynomial fit of degree {degree}')
# Fitted curve
plt.title('Polynomial Fit')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```

```python
# Mean Squared Error (MSE)
mse = np.mean((y - y_pred)**2)
print(f'Mean Squared Error: {mse:.4f}')
# R-squared value
ss_total = np.sum((y - np.mean(y))**2)
ss_res = np.sum((y - y_pred)**2)
r_squared = 1 - (ss_res / ss_total)
print(f'R-squared: {r_squared:.4f}')
```

# Challenge 10a

### Least Squares Regression Code

```python
# Data points
X = np.sort(np.random.uniform(0,10,20))
print(X)
Y = np.random.uniform(0,10,20)
# Stack the design matrix with a column of ones (for the intercept)
A = np.vstack([X, np.ones(len(X))]).T
# Solve using least squares
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
print(f"Slope: {m}, Intercept: {c}")
t=np.linspace(0,10,20)
```

```python
y=m*t+c
```

### Plot Routine

```python
plt.figure()
plt.title("Least Squares Regression")
plt.plot(X,Y, 'b.')
plt.plot(t,y,'r')
plt.show()
```

### MSE and R^2 Calculations

```python
mse = np.mean((Y - y)**2)
print(f'Mean Squared Error: {mse:.4f}')
# R-squared value
ss_total = np.sum((Y - np.mean(Y))**2)
ss_res = np.sum((Y - y)**2)
r_squared = 1 - (ss_res / ss_total)
print(f'R-squared: {r_squared:.4f}')
```

### Degree 3 Polynomial Fit

```python
degree = 3 # Degree of the polynomial
coefficients = np.polyfit(X, Y, degree)
# Print the polynomial coefficients
print("Polynomial Coefficients: ", coefficients)
# Step 3: Use the coefficients to predict y-values (polynomial function)
# Generate polynomial function based on the coefficients
polynomial = np.poly1d(coefficients)
```

### Plot Routine

```python
# Generate predicted y-values based on the fitted polynomial
y_pred = polynomial(X)
# Step 4: Plotting the results
plt.scatter(X, Y, color='blue', label='Data points') # Original data points
plt.plot(X, y_pred, color='red', label=f'Polynomial fit of degree {degree}')
# Fitted curve
plt.title('Polynomial Fit')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```

### Error Calculation

```python
# Mean Squared Error (MSE)
mse = np.mean((Y - y_pred)**2)
print(f'Mean Squared Error: {mse:.4f}')
# R-squared value
ss_total = np.sum((Y - np.mean(Y))**2)
ss_res = np.sum((Y - y_pred)**2)
r_squared = 1 - (ss_res / ss_total)
print(f'R-squared: {r_squared:.4f}')
```

### N Degree Vander Matrix
Solved for polynomial fit

```python
degree=9
V = np.vander(Y, N=degree+1, increasing=True)
# V.T * V
VT_V = np.dot(V.T, V)
# V.T * y
VT_y = np.dot(V.T, y)
# Solve for a (coefficients)
coefficients = np.flip(np.linalg.solve(VT_V, VT_y))
polynomial = np.poly1d(coefficients)
```

### Plot Routine

```python
# Generate predicted y-values based on the fitted polynomial
y_pred = polynomial(X)
# Step 4: Plotting the results
plt.scatter(X, Y, color='blue', label='Data points') # Original data points
plt.plot(X, y_pred, color='red', label=f'Polynomial fit of degree {degree}')
# Fitted curve
plt.title('Polynomial Fit')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```
