```python
import random
import math
import numpy as np
import matplotlib.pyplot as plt
def randomShots(numSamples):
    countTotal=0
    countCircle=0
    for i in range(0,numSamples):
        x=random.uniform(-1,1)
        y=random.uniform(-1,1)
        countTotal=countTotal+1
        if (math.sqrt(x*x+y*y)<1):
            countCircle=countCircle+1
    PI=4*(countCircle/countTotal)
    return PI
def errorCalc(PI):
    absolute=np.abs(np.pi-PI)
    relative=np.abs((np.pi-PI)/np.pi)
    return absolute, relative
n=100
samples=1000
t=np.arange(1,101)
x=np.zeros(n)
y=np.zeros(n)
for i in range(0,n):
    PI=randomShots(samples)
    abs,rel=errorCalc(PI)
    x[i]=abs
    y[i]=rel
plt.figure()
plt.plot(t,x,'r-',label="absolute error")
plt.plot(t,y,'b--',label="relative error")
plt.legend()
plt.show()

    

    
    
            
            
    
```
