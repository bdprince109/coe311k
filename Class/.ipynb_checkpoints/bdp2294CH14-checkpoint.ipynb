# Comparing Forward, Backward, and Central Difference Methods

### Code F(x)=sin(x)

```python
import numpy as np
import matplotlib.pyplot as plt
# Define the function and the domain
f = lambda x: np.sin(x)
a, b = 0, 2 * np.pi # Interval
N = 50 # Number of sample points
# Create a uniform grid
x = np.linspace(a, b, N)
y = f(x)
# Plot the sampled points
plt.plot(x, y, 'o-', label='Sampled Points')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Uniform Sampling of f(x) = sin(x)')
plt.legend()
plt.show()
```

### Derivative Approximation Code

```python
# Step size
h = (b - a) / (N - 1)
# Central Difference Approximation for the derivative
cdf_dx = (f(x + h) - f(x - h)) / (2 * h)
#Forward Difference Approximation for the derivative
fdf_dx=(f(x+h)-f(x))/h
#Backwards Difference Approximation for the derivative
bdf_dx=(f(x)-f(x-h))/h
```

### Plot Routine

```python
plt.figure()
plt.subplot(2,2,1)
plt.plot(x, np.cos(x), 'r-',label="Cos(x)")
plt.plot(x,cdf_dx, 'g-',label="Central Difference")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.subplot(2,2,2)
plt.plot(x,fdf_dx, 'b-',label="Forward Difference")
plt.plot(x, np.cos(x), 'r-',label="Cos(x)")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.subplot(2,2,3)
plt.plot(x,bdf_dx, 'y-',label="Backward Difference")
plt.plot(x, np.cos(x), 'r-',label="Cos(x)")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.subplot(2,2,4)
plt.suptitle("Derivative Approximations Comparison")
plt.plot(x, np.cos(x), 'r-',label="Cos(x)")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.subplots_adjust(wspace=.5, hspace=.5)
plt.show()
```
