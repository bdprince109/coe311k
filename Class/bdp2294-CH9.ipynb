## Class Example

```python
import numpy as np
import matplotlib.pyplot as plt
# Data points
X = np.array([1, 2, 3, 4, 5])
Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])
# Stack the design matrix with a column of ones (for the intercept)
A = np.vstack([X, np.ones(len(X))]).T
# Solve using least squares
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
print(f"Slope: {m}, Intercept: {c}")
t=np.linspace(0,5,20)
y=.95*t+.59
```

## Plot Routine

```python
plt.figure()
plt.title("Least Squares Regression")
plt.plot(X,Y, 'b.')
plt.plot(t,y,'r')
plt.show()
```

## Linear Least Squares

```python
import numpy as np
import matplotlib.pyplot as plt
# Data points
X = np.linspace(0,1,101)
print(X)
Y = 4*X+15+np.random.uniform(-1,1,101)
# Stack the design matrix with a column of ones (for the intercept)
A = np.vstack([X, np.ones(len(X))]).T
# Solve using least squares
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
print(f"Slope: {m}, Intercept: {c}")
t=np.linspace(0,1,101)
y= 3.9469435487198554*t+15.042585767456602
```

```python
plt.figure()
plt.title("Least Squares Regression")
plt.plot(X,Y, 'b.')
plt.plot(t,y,'r')
plt.show()
r=np.abs(Y-y)
rindex=np.arange(0,r.size,1)
print(r)
print(min(r))
print(max(r))
print(rindex[r-.0001<min(r)])
print(rindex[r+.0001>max(r)])
```

The minimum residual (.003) occurs at the X=.73, while the maximum residual (1.004) occurs at X=.1
