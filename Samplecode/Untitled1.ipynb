```python
v_add=v1+v2
v_sub=v1-v2
```

```python
print(v_add,v_sub)
```

```python
dot_prod=np.dot(v1,v2)
```

```python
print(dot_prod)
```

```python
A=np.array([[2,3],[4,-1]])
b=np.array([5,3])
x=np.linalg.solve(A,b)
print(x)
```

```python
A=np.array([[2,1],[1,2]])
e_val,e_vec=np.linalg.eig(A)
print("eiganvalue:",e_val)
print("eiganvector", e_vec)
```

```python
A=np.array([[4,7],[2,6]])
A_inv=np.linalg.inv(A)
print(A)
print(A_inv)
```

```python
identity=np.dot(A,A_inv)
print(identity)
```
