## Taylor series function:

```python
import numpy as np
import matplotlib.pyplot as plt
def taylorSin(terms, x):
    sin=0
    for i in range(0,terms):
        sin=sin+((-1)**i)*(x**(2*i+1))/(np.math.factorial(2*i+1))
    return sin
```

## Error calc function:

```python
def errorCalc(estimate,x):
    absolute=np.abs(estimate-np.sin(x))
    return absolute
```

## Plot Routine

```python
x=np.arange(0,10.1,.1)
y1=errorCalc(taylorSin(1,x),x)
y2=errorCalc(taylorSin(2,x),x)
y3=errorCalc(taylorSin(3,x),x)
y4=errorCalc(taylorSin(4,x),x)

plt.figure(figsize=(10, 10))
plt.subplot(2,2,1)
plt.suptitle('Absolute Error of Taylor Series for Sin(x)', y=.98, fontsize=20)
plt.plot(x,y1,'r')
plt.title('One Term',fontsize=15)
plt.subplot(2,2,2)
plt.plot(x,y2,'b')
plt.title('Two Terms',fontsize=15)
plt.subplot(2,2,3)
plt.plot(x,y3,'g')
plt.title('Three Terms',fontsize=15)
plt.subplot(2,2,4)
plt.plot(x,y4,'y')
plt.title('Four Terms',fontsize=15)
plt.subplots_adjust(hspace=.25, wspace=.25)
plt.show()
```

The truncation error is smaller over the first few values of x with the addition of more terms, however once the x value exceeds the bounds of where the taylor series is well fit to sin(x), the truncation error increases much more rapidly for those taylor series approximations with more terms compared to a taylor series with less terms. This means that a taylor series with more terms is only more accurate up to a point after which it becomes wildly inaccurate.
